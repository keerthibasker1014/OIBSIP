```python
import pandas as pd
import numpy as np

print("Libraries imported successfully")
```

```python
import os

print(os.listdir('/content'))
```

```python
df = pd.read_csv('/content/train.csv')

print("Dataset loaded successfully!")
print("shape:",df.shape)

display(df.head())
```

```python
print("Missing values per column:")
display(df.isnull().sum())

print("Duplicate row:")
print(df.duplicated().sum())

print("\nData types:")
display(df.dtypes)

print("\nDataset information:")
df.info()
```

```python
missing = df.isnull().sum()

print("Missing values before cleaning:")
display(missing[missing > 0])
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

df = df.drop(columns=['Cabin'])

print("Missing values after cleaning:")
display(df.isnull().sum())
```

I checked the missing values in the dataset and handles them based on the data.

```python
duplicates = df.duplicated().sum()

print("Duplicate rows before cleaning:",duplicates)
df = df.drop_duplicates()

print("Duplicate rows after cleaning:",df.duplicated().sum())
```

I checked the dataset for duplicate rows and removed them to avoid repeated recors.This hepls to keep the dataset clean and give accurate analysis

```python
print("Sex values:")
print(df['Sex'].unique())

print("\nEmbarked values:")
print(df['Embarked'].unique())

```

```python
df['Sex'] = df['Sex'].str.strip().str.lower()
df['Embarked'] = df['Embarked'].str.strip().str.upper()

print("Standardisation completed.")
print(df['Sex'].unique())
print(df['Embarked'].unique())
```

I checked the values and made their format.
this hepls avoid prblems caused by different or extra spaces in the data

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Age'] < lower_bound) | (df['Age'] > upper_bound)]

print("Number of Age outliers:",len(outliers))

```

I used the IQR method to identify possible outliers in the age column.

```python
print("Data types before correction:")
print(df.dtypes)
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Fare'] = pd.to_numeric(df['Fare'], errors='coerce')
df['Sex'] = df['Sex'].astype('string')
df['Embarked'] = df['Embarked'].astype('string')

print("Data types after correction:")
print(df.dtypes)
```

I checked the data types and corrected the important columns.This makes the dataset easier to analyse

```python
print("Final Dataset Summary")
print("________________")
print("Rows:",df.shape[0])
print("Columns:",df.shape[1])
print("Missing Values:",df.isnull().sum().sum())
print("\nFinal data types:")
print(df.dtypes)
```

After cleaning the dataset,the missing values were handled,duplicate rows were removed,categorical values were standarised,and data type were corrected.
The cleaned dataset is now ready for analysis.

```python
df.to_csv('Cleaned_titanic.csv',index=False)
print("Cleaned dataset saved successfully")
```
